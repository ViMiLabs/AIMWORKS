# H₂KG-backed characterization knowledge graph
Load the named RDF graphs, execute the published SPARQL competency queries, and validate the flattened graph with pySHACL.


```python
from pathlib import Path
from rdflib import Dataset, Graph
from pyshacl import validate
ROOT = Path('.')
dataset = Dataset()
dataset.parse(ROOT / 'characterization_knowledge_graph.trig', format='trig')
len(dataset)
```

```python
query = (ROOT / 'queries' / 'q1_tem_specimen_preparation.rq').read_text(encoding='utf-8')
list(dataset.query(query))
```

```python
data = Graph().parse(ROOT / 'characterization_knowledge_graph.ttl', format='turtle')
shapes = Graph().parse(ROOT / 'characterization_shapes.ttl', format='turtle')
validate(data, shacl_graph=shapes, inference='rdfs')[0]
```
